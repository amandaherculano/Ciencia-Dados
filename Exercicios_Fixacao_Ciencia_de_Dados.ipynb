## Exercício 2 — Array numpy: média, desvio padrão, soma e centralização

**Enunciado:** Em Python, crie um array numpy com `[10, 20, 30, 40, 50]`. Calcule média, desvio
padrão e soma. Depois centralize o array (subtraia a média) e verifique que a nova média é zero.

````python
import numpy as np

x = np.array([10, 20, 30, 40, 50])

media = x.mean()
desvio_padrao = x.std()
soma = x.sum()

print("Array original:", x)
print("Média:", media)
print("Desvio padrão:", desvio_padrao)
print("Soma:", soma)

# Centralizar o array (subtrair a média de cada elemento)
x_centralizado = x - media
print("\nArray centralizado:", x_centralizado)
print("Nova média (deve ser ~0):", x_centralizado.mean())
````

**Observação:** a nova média não é exatamente `0` por causa de arredondamento de ponto
flutuante (algo como `-1.42e-16`), mas é numericamente igual a zero. Isso é esperado sempre
que se trabalha com aritmética de ponto flutuante em vez de números reais exatos.

## Exercício 3 — DataFrame pandas: filtro, seleção de colunas e `loc` vs `iloc`

**Enunciado:** Em Python, crie um DataFrame com colunas `nome`, `idade` e `nota` (ao menos cinco
linhas). Calcule a nota média. Filtre os alunos com nota acima de 7 e mostre apenas as colunas
`nome` e `nota` desse subconjunto. Depois use `loc` e `iloc` para acessar a primeira linha do
DataFrame original; explique a diferença por escrito em um comentário.

````python
import pandas as pd

df = pd.DataFrame({
    "nome": ["Ana", "Bruno", "Carla", "Diego", "Elisa"],
    "idade": [23, 30, 27, 22, 25],
    "nota": [8.5, 6.0, 9.2, 7.8, 5.5]
})

df
````

````python
# Nota média de toda a turma
nota_media = df["nota"].mean()
print("Nota média:", nota_media)

# Filtrar alunos com nota acima de 7, mostrando apenas nome e nota
aprovados = df[df["nota"] > 7][["nome", "nota"]]
aprovados
````

````python
# Acessando a primeira linha do DataFrame original com loc e iloc
linha_loc = df.loc[0]   # loc acessa por RÓTULO do índice
linha_iloc = df.iloc[0] # iloc acessa por POSIÇÃO (independente do rótulo)

print(linha_loc)
print()
print(linha_iloc)

# Comentário explicativo:
# Como o índice do df ainda é o padrão (0, 1, 2, 3, 4), loc[0] e iloc[0]
# retornam o mesmo resultado neste caso específico. A diferença aparece
# quando o índice não é sequencial/numérico simples (por exemplo, após um
# filtro ou um sort que preserva os rótulos originais): loc continua
# buscando pelo RÓTULO 0, enquanto iloc sempre busca pela PRIMEIRA
# POSIÇÃO física da tabela, não importa qual rótulo ela tenha.
````

````python
# Exemplo que evidencia a diferença entre loc e iloc:
# depois de filtrar, os rótulos originais do índice são preservados
df_ordenado = df.sort_values("nota", ascending=False)
print(df_ordenado)

print("\nloc[0] busca a linha ROTULADA 0 (a da Ana), mesmo após reordenar:")
print(df_ordenado.loc[0])

print("\niloc[0] busca a PRIMEIRA linha na ordem atual (a de maior nota):")
print(df_ordenado.iloc[0])
````

## Exercício 4 — O mesmo tibble em R: `summarise` e pipeline com `filter`/`select`

**Enunciado:** Em R, recrie o mesmo tibble. Calcule a nota média usando `summarise`. Depois,
encadeando com pipe, filtre alunos com nota acima de 7 e selecione só as colunas `nome` e
`nota`. Comente em uma linha como você faria a mesma sequência em Python usando pandas.

**Código em R** (rodar no RStudio, conforme Seção 3 e 8 da apostila):

```r
library(tidyverse)

df <- tibble(
  nome  = c("Ana", "Bruno", "Carla", "Diego", "Elisa"),
  idade = c(23, 30, 27, 22, 25),
  nota  = c(8.5, 6.0, 9.2, 7.8, 5.5)
)

# Nota média com summarise
df |>
  summarise(nota_media = mean(nota))

# Pipeline: filtrar nota > 7 e selecionar nome e nota
df |>
  filter(nota > 7) |>
  select(nome, nota)
```

**Comentário comparativo (uma linha):** em pandas, a mesma sequência seria
`df[df["nota"] > 7][["nome", "nota"]]` — o `filter()` do dplyr equivale à indexação booleana
`df[condição]`, e o `select()` equivale a escolher colunas com uma lista `[["col1", "col2"]]`,
só que o pipe `|>` do R encadeia as chamadas na ordem de leitura, enquanto em pandas o
encadeamento fica "de dentro para fora" (ou é feito com `.query()` + `.filter()`/colchetes
em sequência, como já demonstrado na célula de código do Exercício 3).

## Exercício 6 — Contagem de pinguins por espécie e por ilha (Python x R)

**Enunciado:** Descubra quantos pinguins existem por espécie, e depois por ilha. Compare a
contagem obtida em Python com a obtida em R.

**Em R** (Seção 9.2 da apostila):

```r

count(penguins, island)
```

**Comparação:** os números obtidos em Python (`value_counts()`) e em R (`count()`) devem
coincidir, pois é o mesmo dataset. Conforme já adiantado na Seção 9.3 da apostila, a
distribuição esperada por espécie é:

- Adélie: 152
- Gentoo: 124
- Chinstrap: 68

E a distribuição por ilha também é desigual, o que reforça a observação de que **as espécies
não estão igualmente distribuídas entre as ilhas** — ponto que será retomado na análise
